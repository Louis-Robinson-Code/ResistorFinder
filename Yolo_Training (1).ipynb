```python
from ultralytics import YOLO
import matplotlib.pyplot as plt
```

```python
#selecting and training the model
model =YOLO('yolo11s.pt')  #selecting our model to be yolo v11s
model.train(
    data='/content/custom_data/data.yaml',  # yamal file containing all the classes
    epochs=80,                  # epochs, 80 chosen to try optimise accuracy and execution time
    imgsz=640                   # image size required for yolo v11s
)
```

```python
import cv2
model_path = "/content/runs/detect/train/weights/best.pt"
test_img =  "/content/5B-2K2-T1-3_jpg.rf.ec66864fd3de7d03fc9e8d44e181b304.jpg"
yolo_model = YOLO(model_path) #setting up the model
results = yolo_model.predict(test_img, save=True)#inputing the testimg in to get the band values
for result in results: #accessing the results
  img_array = result.plot()  # Returning as numpy arr

img_array = cv2.cvtColor(img_array, cv2.COLOR_BGR2RGB) #convert back to RGB as yolo returns BGR
plt.imshow(img_array)
```
